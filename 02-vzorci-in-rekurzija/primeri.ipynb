# Vzorci in rekurzija

## Definicija `let rec`

````ocaml
let rec vsota_prvih n =
  if n = 0 then 0 else vsota_prvih (n - 1) + n
````

````ocaml
vsota_prvih 100
````

````ocaml
let rec fakulteta n =
  if n = 0 then 1 else n * fakulteta (n - 1)
````

````ocaml
fakulteta 10
````

````ocaml
let rec fib n =
  if n = 0 then 0
  else if n = 1 then 1
  else fib (n - 1) + fib (n - 2)
````

````ocaml
let hitri_fib n =
  let rec aux n a b =
    if n = 0 then a else aux (n - 1) b (a + b)
  in aux n 0 1
````

## Konstrukta `match` in `function`

````ocaml
let rec vsota_prvih n =
  match n with
  | 0 -> 0
  | _ -> vsota_prvih (n - 1) + n
````

````ocaml
vsota_prvih 100
````

````ocaml
let rec fakulteta n =
  match n with
  | 0 -> 1
  | _ -> n * fakulteta (n - 1)
````

````ocaml
let rec fib n =
  match n with
  | 0 -> 0
  | 1 -> 1
  | _ -> fib (n - 1) + fib (n - 2)
````

````ocaml
let rec vsota_prvih =
  function
  | 0 -> 0
  | n -> vsota_prvih (n - 1) + n
````

## Razstavljanje seznamov

````ocaml
let citiraj_knjigo avtorji naslov =
  match avtorji with
  | [] -> naslov
  | [avtor] -> avtor ^ ": " ^ naslov
  | [prvi; drugi] -> prvi ^ ", " ^ drugi ^ ": " ^ naslov
  (* uporaba List.hd je slaba praksa *)
  | _ -> List.hd avtorji ^ " in ostali: " ^ naslov
````

````ocaml
List.hd []
````

````ocaml
citiraj_knjigo [] "Telefonski imenik"
````

````ocaml
citiraj_knjigo ["Tolkien"] "Gospodar prstanov"
````

````ocaml
citiraj_knjigo ["Pratchett"; "Gaiman"] "Dobra znamenja"
````

````ocaml
citiraj_knjigo ["Golob"; "Kos"; "Vrabec"] "Ptiči Slovenije"
````

## Rekurzivne funkcije na seznamih

````ocaml
let rec vsota =
  function
  | [] -> 0
  | x :: xs -> x + vsota xs
````

````ocaml
vsota [1; 2; 3; 4]
````

````ocaml
let rec dolzina =
  function
  | [] -> 0
  | _ :: xs -> 1 + dolzina xs
````

````ocaml
dolzina [1; 2; 3; 4]
````

````ocaml
let rec map f xs =
  match xs with
  | [] -> []
  | x :: xs' -> f x :: map f xs'
````

````ocaml
map String.length ["avto"; "pes"; "mačka"]
````

````ocaml
let rec ( @ ) sez1 sez2 =
  match sez1 with
  | [] -> sez2
  | x :: sez1' -> x :: (sez1' @ sez2)
````

````ocaml
[1; 2; 3] @ [4; 5; 6]
````

## Repna rekurzija

````ocaml
let rec harmonicna =
  function
  | 0 -> 0.
  | n -> 1. /. float_of_int n +. harmonicna (n - 1)
````

````ocaml
harmonicna 260000
````

````ocaml
let rec harmonicna' acc n =
  match n with
  | 0 -> acc
  | _ -> harmonicna' (1. /. float_of_int n +. acc) (n - 1)
````

````ocaml
harmonicna 260000
````

````ocaml
harmonicna' 0. 100000000
````

````ocaml
let rec vsota sez =
  match sez with
  | [] -> 0
  | x :: xs -> x + vsota xs

let vsota' sez =
  let rec aux delna_vsota =
    function
    | [] -> delna_vsota
    | glava :: rep -> aux (delna_vsota + glava) rep
  in
  aux 0 sez
````

````ocaml
vsota' (List.init 1000000 (fun x -> x))
````

````ocaml
let produkt seznam =
  let rec aux acc =
    function
    | [] -> acc
    | x :: xs -> aux (x * acc) xs
  in
  aux 1 seznam
  
let map f seznam =
  let rec aux acc =
    function
    | [] -> acc
    | x :: xs -> aux (acc @ [f x]) xs
  in
  aux [] seznam
  
````

````ocaml
map succ [1; 2; 3; 4; 1]
````

````ocaml
let map' f seznam =
  let rec aux acc =
    function
    | [] -> List.rev acc
    | glava :: rep -> aux (f glava :: acc) rep
  in
  aux [] seznam
````

````ocaml
map' (fun x -> 10 * x) [1; 2; 3; 4]
````

## Zlaganje seznamov

````ocaml
let rec fold_right f xs z =
  match xs with
  | [] -> z
  | x :: xs -> f x (fold_right f xs z)
````

````ocaml
fold_right ( + ) [1; 2; 3; 4] 0
````

````ocaml
fold_right ( * ) [1; 2; 3; 4] 1
````

````ocaml
let rec fold_left f z xs =
  match xs with
  | [] -> z
  | x :: xs -> fold_left f (f z x) xs
````

````ocaml
List.fold_left (^) "X" ["A"; "B"; "C"]
````

````ocaml
List.fold_left (+) 0 [10; 20; 30]
````

````ocaml
List.fold_left (fun xs x -> x :: xs) [] [1; 2; 3]
````

````ocaml
let rec potenca a n =
  match n with
  | 0 -> 1
  | n -> a * potenca a (n - 1)
````

````ocaml
List.fold_right potenca [2; 3; 2] 1
````

````ocaml
List.fold_left potenca 2 [3; 2]
````

```python
def fold_left(f, z, xs):
    acc = z
    for x in xs:
        acc = f(acc, x)
    return acc
```

````ocaml
let rec vstavi x =
  function
  | [] -> [x]
  | glava :: rep when x <= glava -> x :: glava :: rep
  | glava :: rep -> glava :: vstavi x rep
````

````ocaml
vstavi 3 [1; 2; 4; 5]
````
