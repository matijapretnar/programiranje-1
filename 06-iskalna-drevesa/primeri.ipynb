# Iskalna drevesa

```ocaml
module type MNOZICA = sig
    type 'a t
    val prazna : 'a t
    val dodaj : 'a -> 'a t -> 'a t
    val velikost : 'a t -> int
    val vsebuje : 'a -> 'a t -> bool
  end
```

```ocaml
module MnozicaPrekSeznamov : MNOZICA = struct
    type 'a t = 'a list
    let prazna = []
    let velikost m = List.length m
    let vsebuje x m = List.mem x m
    let dodaj x m = if vsebuje x m then m else x :: m
  end
```

```ocaml
module StetjeRazlicnih (M : MNOZICA) = struct
    let stevilo_razlicnih xs =
      let rec aux ze_videni = function
        | [] -> M.velikost ze_videni
        | x :: xs -> aux (M.dodaj x ze_videni) xs
      in
      aux M.prazna xs
  end
```

```ocaml
module S = StetjeRazlicnih (MnozicaPrekSeznamov)
```

```ocaml
S.stevilo_razlicnih [1; 2; 1; 2; 1; 2; 3]
```

```ocaml
let nakljucni_seznam m n = List.init n (fun _ -> Random.int m)
```

```ocaml
let stopaj f x =
    let zacetek = Sys.time () in
    let y = f x in
    let konec = Sys.time () in
    let izpis =
      Printf.sprintf "Porabljen čas: %f ms" (1000. *. (konec -. zacetek))
    in
    print_endline izpis;
    y
```

```ocaml
stopaj S.stevilo_razlicnih (nakljucni_seznam 1_000_000_000 4000)
```

```ocaml
stopaj S.stevilo_razlicnih (nakljucni_seznam 1_000_000_000 8000)
```

```ocaml
module IskalnoDrevo = struct
    type 'a t = 'a drevo

    let prazna = Prazno

    let rec velikost = function
      | Prazno -> 0
      | Sestavljeno (l, _, d) -> 1 + velikost l + velikost d

    let rec vsebuje x =
      function
      | Prazno -> false
      | Sestavljeno (l, y, d) when x = y -> true
      | Sestavljeno (l, y, d) ->
          if x < y then vsebuje x l else vsebuje x d

    let rec dodaj x =
      function
      | Prazno -> Sestavljeno (Prazno, x, Prazno)
      | (Sestavljeno (_, y, _) as drevo) when x = y -> drevo
      | Sestavljeno (l, y, d) ->
          if x < y then
            Sestavljeno (dodaj x l, y, d)
          else
            Sestavljeno (l, y, dodaj x d)
  end
```

## AVL drevesa

```ocaml
module AVLDrevo = struct
    include IskalnoDrevo

    let razlika = function
      | Prazno -> 0
      | Sestavljeno (l, _, d) -> visina l - visina d

    let zavrti_levo = function
      | Sestavljeno (l, x, Sestavljeno (dl, y, dd)) ->
          Sestavljeno (Sestavljeno (l, x, dl), y, dd)
      | _ -> failwith "Tega drevesa ne morem zavrteti"

    let zavrti_desno = function
      | Sestavljeno (Sestavljeno (ll, y, ld), x, d) ->
          Sestavljeno (ll, y, Sestavljeno (ld, x, d))
      | _ -> failwith "Tega drevesa ne morem zavrteti"

    let uravnotezi drevo =
      match drevo with
      | Sestavljeno (l, _, _) when razlika drevo = 2 && razlika l = 1 ->
          zavrti_desno drevo
      | Sestavljeno (l, x, d) when razlika drevo = 2 ->
          Sestavljeno (zavrti_levo l, x, d) |> zavrti_desno
      | Sestavljeno (_, _, d) when razlika drevo = -2 && razlika d = -1 ->
          zavrti_levo drevo
      | Sestavljeno (l, x, d) when razlika drevo = -2 ->
          Sestavljeno (l, x, zavrti_desno d) |> zavrti_levo
      | _ -> drevo

    let rec dodaj x drevo =
      match drevo with
      | Prazno -> Sestavljeno (Prazno, x, Prazno)
      | Sestavljeno (l, y, d) ->
          if x < y then Sestavljeno (dodaj x l, y, d) |> uravnotezi
          else if x > y then Sestavljeno (l, y, dodaj x d) |> uravnotezi
          else drevo
  end
```

```ocaml
module AVLDrevoZVisinami = struct
    type 'a t =
    | Prazno
    | Sestavljeno of int * 'a t * 'a * 'a t

    let prazna = Prazno

    let visina = function
      | Prazno -> 0
      | Sestavljeno (h, _, _, _) -> h

    let sestavljeno (l, x, d) =
        let h = 1 + max (visina l) (visina d) in
        Sestavljeno (h, l, x, d)

    let rec velikost = function
      | Prazno -> 0
      | Sestavljeno (_, l, _, d) -> 1 + velikost l + velikost d

    let rec vsebuje x =
      function
      | Prazno -> false
      | Sestavljeno (_, l, y, d) when x = y -> true
      | Sestavljeno (_, l, y, d) ->
          if x < y then vsebuje x l else vsebuje x d

    let razlika = function
      | Prazno -> 0
      | Sestavljeno (_, l, _, d) -> visina l - visina d

    let zavrti_levo = function
      | Sestavljeno (_, l, x, Sestavljeno (_, dl, y, dd)) ->
          sestavljeno (sestavljeno (l, x, dl), y, dd)
      | _ -> failwith "Tega drevesa ne morem zavrteti"

    let zavrti_desno = function
      | Sestavljeno (_, Sestavljeno (_, ll, y, ld), x, d) ->
          sestavljeno (ll, y, sestavljeno (ld, x, d))
      | _ -> failwith "Tega drevesa ne morem zavrteti"

    let uravnotezi drevo =
      match drevo with
      | Sestavljeno (_, l, _, _) when razlika drevo = 2 && razlika l = 1 ->
          zavrti_desno drevo
      | Sestavljeno (_, l, x, d) when razlika drevo = 2 ->
          sestavljeno (zavrti_levo l, x, d) |> zavrti_desno
      | Sestavljeno (_, _, _, d) when razlika drevo = -2 && razlika d = -1 ->
          zavrti_levo drevo
      | Sestavljeno (_, l, x, d) when razlika drevo = -2 ->
          sestavljeno (l, x, zavrti_desno d) |> zavrti_levo
      | _ -> drevo

    let rec dodaj x drevo =
      match drevo with
      | Prazno -> sestavljeno (Prazno, x, Prazno)
      | Sestavljeno (_, l, y, d) ->
          if x < y then sestavljeno (dodaj x l, y, d) |> uravnotezi
          else if x > y then sestavljeno (l, y, dodaj x d) |> uravnotezi
          else drevo
  end
```

```ocaml
module MnozicaPrekAVLDrevesZVisinami : MNOZICA = AVLDrevoZVisinami
```

```ocaml
module S = StetjeRazlicnih (MnozicaPrekAVLDrevesZVisinami)
```
