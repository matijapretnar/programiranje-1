## Urejanje z vstavljanjem

```ocaml
let rec vstavi x xs =
    match xs with
    | [] -> [x]
    | x' :: xs' -> if x <= x' then x :: x' :: xs' else x' :: vstavi x xs'

let rec vstavi x xs =
    match xs with
    | x' :: xs' when x > x' -> x' :: vstavi x xs'
    | _ -> x :: xs

let rec uredi_z_vstavljanjem xs =
    match xs with
    | [] -> []
    | x :: xs' -> vstavi x (uredi_z_vstavljanjem xs')
```

```ocaml
uredi_z_vstavljanjem [3; 2; 5; 9; 1; 7; 4; 3]
```

```ocaml
let nakljucni_seznam m n = List.init n (fun _ -> Random.int m)
```

```ocaml
let stopaj f x =
    let zacetek = Sys.time () in
    let y = f x in
    let konec = Sys.time () in
    let izpis =
      Printf.sprintf "Porabljen čas: %f ms" (1000. *. (konec -. zacetek))
    in
    print_endline izpis;
    y
```

```ocaml
[2_000; 4_000; 8_000; 16_000]
  |> List.map (nakljucni_seznam 1_000_000_000)
  |> List.iter (stopaj (fun sez -> ignore @@ uredi_z_vstavljanjem sez))
```

```ocaml
[200_000; 400_000; 800_000; 1600_000]
  |> List.map (nakljucni_seznam 1_000_000_000)
  |> List.iter (stopaj (fun sez -> ignore @@ List.sort compare sez))
```

## Notaciji $\Omega$ in $\Theta$

$$f \in \Omega(g) \iff g \in O(f) \iff \exists c > 0, n_0.\ \forall n \ge n_0.\ f(n) \ge c \cdot g(n)$$

$$f \in \Theta(g) \iff f \in O(g) \land f \in \Omega(g) \iff \exists c, C > 0, n_0. \forall n \ge n_0.\  c \cdot g(n) \le f(n) \le C \cdot g(n)$$

## Pravila za računanje

$$f \in O(g) \implies c \cdot f \in O(g)$$

$$f_1 \in O(g_1) \land f_2 \in O(g_2) \implies f_1 + f_2 \in O(g_1 + g_2)$$

$$f_1 \in O(g_1) \land f_2 \in O(g_2) \implies f_1 \cdot f_2 \in O(g_1 \cdot g_2)$$

$$O_{C,n_0}(g) = \{ f \mid \forall n \ge n_0.\ f(n) \le C \cdot g(n) \}$$

```ocaml
let obrni sez =
    let rec aux acc = function
      | [] -> acc
      | glava :: rep -> aux (glava :: acc) rep
    in
    aux [] sez
```

```ocaml
let s = [1; 5; 10; 20]
```

```ocaml
obrni s
```

```ocaml
s
```

```ocaml
let vrni_obrnjeno tabela =
    let n = Array.length tabela in
    Array.init n (fun i -> tabela.(n - i - 1))
```

```ocaml
let t = [|1; 5; 10; 20|]
```

```ocaml
vrni_obrnjeno t
```

```ocaml
t
```

```ocaml
let zamenjaj tabela i j =
    let t = tabela.(i) in
    tabela.(i) <- tabela.(j);
    tabela.(j) <- t
```

```ocaml
let obrni_na_mestu tabela =
    let rec aux i j =
      if i < j then (
        zamenjaj tabela i j;
        aux (i + 1) (j - 1))
    in
    aux 0 (Array.length tabela - 1)
```

```ocaml
let t' = [|3; 33; 333; 3333|]
```

```ocaml
obrni_na_mestu t'
```

```ocaml
let rec razdeli_na_pol = function
    | [] -> ([], [])
    | [x] -> ([x], [])
    | x1 :: x2 :: xs ->
        let (xs1, xs2) = razdeli_na_pol xs in
        (x1 :: xs1, x2 :: xs2)
```

```ocaml
razdeli_na_pol [1; 2; 3; 4; 5; 6; 7; 8]
```

```ocaml
let rec zlij xs ys =
    match (xs, ys) with
    | ([], _) -> ys
    | (_, []) -> xs
    | (x :: xs', y :: ys') ->
        if x <= y then
            x :: zlij xs' ys
        else
            y :: zlij xs ys'
```

```ocaml
zlij [1; 10; 11; 100; 101; 110; 111] [2; 20; 22; 200; 202; 220; 222]
```

```ocaml
let rec uredi_z_zlivanjem sez =
    match sez with
    | [] | [_] -> sez
    | _ ->
        let (sez1, sez2) = razdeli_na_pol sez in
        zlij (uredi_z_zlivanjem sez1) (uredi_z_zlivanjem sez2)
```

```ocaml
uredi_z_zlivanjem [3; 2; 5; 9; 1; 7; 4; 3]
```

```ocaml
let pivotiraj p xs =
    List.partition (fun x -> x <= p) xs
```

```ocaml
pivotiraj 3 [2; 5; 9; 1; 7; 4; 3]
```

```ocaml
let rec hitro_uredi sez =
    match sez with
    | [] | [_] -> sez
    | p :: xs ->
      let sez1, sez2 = pivotiraj p xs in
      let sez1' = hitro_uredi sez1
      and sez2' = hitro_uredi sez2 in
      sez1' @ p :: sez2'
```

```ocaml
hitro_uredi [3; 2; 5; 9; 1; 7; 4; 3]
```

```ocaml
let pivotiraj_na_mestu tabela zacetek konec =
    let pivot = tabela.(zacetek) in
    let rec aux i zacetek_vecjih =
      if i > konec then zacetek_vecjih
      else if tabela.(i) <= pivot then (
        zamenjaj tabela i zacetek_vecjih;
        aux (i + 1) (zacetek_vecjih + 1))
      else aux (i + 1) zacetek_vecjih
    in
    let p = aux (zacetek + 1) (zacetek + 1) - 1 in
    zamenjaj tabela zacetek p;
    p
```

```ocaml
let hitro_uredi_na_mestu tabela =
    let rec uredi_med_indeksoma i j =
      if i < j then
        let p = pivotiraj_na_mestu tabela i j in
        uredi_med_indeksoma i (p - 1);
        uredi_med_indeksoma (p + 1) j
    in
    uredi_med_indeksoma 0 (Array.length tabela - 1)
```

```ocaml
let t = [|3; 2; 5; 9; 1; 7; 4; 3|]
```

```ocaml
hitro_uredi_na_mestu t
```

```ocaml
let tabela_nakljucnih m n = Array.init n (fun _ -> Random.int m)
```

```ocaml
[125_000; 250_000; 500_000; 1_000_000]
  |> List.map (tabela_nakljucnih 1_000_000_000)
  |> List.iter (stopaj hitro_uredi_na_mestu)
```

```ocaml
[2_500; 5_000; 10_000]
  |> List.map (fun n -> Array.init n (fun i -> i))
  |> List.iter (stopaj hitro_uredi_na_mestu)
```

```ocaml
let pivotiraj_nakljucno tabela zacetek konec =
    zamenjaj tabela zacetek (zacetek + Random.int (konec - zacetek + 1));
    pivotiraj_na_mestu tabela zacetek konec
```

```ocaml
let hitro_uredi_na_mestu tabela =
    let rec uredi_med_indeksoma i j =
      if i < j then
        let p = pivotiraj_nakljucno tabela i j in
        uredi_med_indeksoma i (p - 1);
        uredi_med_indeksoma (p + 1) j
    in
    uredi_med_indeksoma 0 (Array.length tabela - 1)
```
