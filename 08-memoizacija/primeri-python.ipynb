## Memoizacija s funkcijo višjega reda

```python
def dodaj_spomin(f_brez_spomina):
    spomin = {}
    def f_s_spominom(*args):
        if args not in spomin:
            spomin[args] = f_brez_spomina(*args)
        return spomin[args]
    return f_s_spominom

def stolpi(n):
    print("Računam", n)
    if n < 0:
        return 0
    elif n == 0:
        return 1
    else:
        return stolpi(n - 1) + stolpi(n - 2) + stolpi(n - 3)
```

```python
stolpi_s_spominom = dodaj_spomin(stolpi)
```

```python
stolpi_s_spominom(5)
```

```python
stolpi = dodaj_spomin(stolpi)
```

## Dekoratorji

```python
import time

def dodaj_stopanje(f):
    def f_ki_stopa(*args):
        zacetek = time.time()
        y = f(*args)
        konec = time.time()
        print(f"Računanje funkcije {f.__name__} z argumenti {args} je trajalo {konec - zacetek:.6f} sekund.")
        return y
    return f_ki_stopa
```

```python
@dodaj_stopanje
def fakulteta(n):
    produkt = 1
    for i in range(1, n + 1):
        produkt *= i
    return produkt
```

## Najcenejša pot v matriki

```python
matrika = [
    [131, 673, 234, 103, 18],
    [201, 96, 342, 965, 150],
    [630, 803, 746, 422, 111],
    [537, 699, 497, 121, 956],
    [805, 732, 524, 37, 331],
]
```

```python
def cena_najcenejse_poti(matrika):
    m, n = len(matrika), len(matrika[0])
    @cache
    def pomozna(i, j):
        polje = matrika[i][j]
        if i == m - 1 and j == n - 1:
            return polje
        elif i == m - 1:
            return polje + pomozna(i, j + 1)
        elif j == n - 1:
            return polje + pomozna(i + 1, j)
        else:
            return polje + min(
                pomozna(i + 1, j),
                pomozna(i, j + 1),
            )
    return pomozna(0, 0)
```

```python
cena_najcenejse_poti(matrika)
```
