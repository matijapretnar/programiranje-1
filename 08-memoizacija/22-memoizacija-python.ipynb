## Funkcija `cache`

```python
def stolpi(n):
    print("Računam", n)
    if n < 0:
        return 0
    elif n == 0:
        return 1
    else:
        return stolpi(n - 1) + stolpi(n - 2) + stolpi(n - 3)
```

## Najcenejša pot v matriki

```python
matrika = [
    [131, 673, 234, 103, 18],
    [201, 96, 342, 965, 150],
    [630, 803, 746, 422, 111],
    [537, 699, 497, 121, 956],
    [805, 732, 524, 37, 331],
]
```

```python
def cena_najcenejse_poti_iz(matrika, i, j):
    m, n = len(matrika), len(matrika[0])
    polje = matrika[i][j]
    if i == m - 1 and j == n - 1:
        return polje
    elif i == m - 1:
        return polje + cena_najcenejse_poti_iz(matrika, i, j + 1)
    elif j == n - 1:
        return polje + cena_najcenejse_poti_iz(matrika, i + 1, j)
    else:
        return polje + min(
            cena_najcenejse_poti_iz(matrika, i + 1, j),
            cena_najcenejse_poti_iz(matrika, i, j + 1),
        )
```
